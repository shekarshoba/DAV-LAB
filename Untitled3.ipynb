```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Iris.csv")
sns.histplot(data=df,x="PetalLengthCm",bins=10)
plt.title("Distribution of petal length")
plt.xlabel("PetalLength(Cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["PetalLengthCm"], bins=10)
plt.title("Distribution of petal length")
plt.xlabel("PetalLength(Cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
df=pd.read_csv("Iris.csv")
plt.boxplot(df["PetalLengthCm"])
plt.title("Iris petal length Box plot {matplotlib)")
plt.ylabel("Length(cm)")
plt.show()
```

```python
sns.boxplot(data=df, y="PetalLengthCm")
plt.title("overall petal length distribution")
plt.show()
```

```python
sns.boxplot(data=df,x="Species", y="PetalLengthCm")
plt.title(" petal length by Iris Species")
plt.show()
```

```python
plt.scatter(df["PetalLengthCm"], df["PetalWidthCm"])
plt.title("petal length vs petal width")
plt.xlabel("petal length(cm)")
plt.ylabel("petal width(cm)")
plt.show()

```

```python
sns.scatterplot(data=df,x="PetalLengthCm",y="PetalWidthCm",hue="Species")
plt.title("petal length vs petal width (seaborn)")
plt.xlabel("petal length(cm)")
plt.ylabel("petal width(cm)")
plt.show()

```

```python
sns.pairplot(df.drop(columns=["Id"]), hue="Species", palette=["red","blue","green"])
plt.show()
```

```python
numerical_df=df.select_dtypes("number").drop(columns=["Id"],errors="ignore")
correlation=numerical_df.corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("correlation Heatmap")
plt.show()
```
